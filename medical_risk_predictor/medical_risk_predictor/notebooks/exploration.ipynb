# Medical Disease Risk Exploration
Exploratory Data Analysis notebook for Diabetes and Heart Disease datasets.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
base = Path('..') / 'data' / 'raw'
diabetes_path = base / 'diabetes.csv'
heart_path = base / 'heart.csv'
diabetes_df = pd.read_csv(diabetes_path) if diabetes_path.exists() else pd.DataFrame()
heart_df = pd.read_csv(heart_path) if heart_path.exists() else pd.DataFrame()
print('Diabetes shape:', diabetes_df.shape)
print('Heart shape:', heart_df.shape)
```

```python
if not diabetes_df.empty:
    display(diabetes_df.head())
    display(diabetes_df.describe(include='all'))
```

```python
if not diabetes_df.empty:
    plt.figure(figsize=(10, 6))
    sns.heatmap(diabetes_df.corr(numeric_only=True), cmap='coolwarm', annot=False)
    plt.title('Diabetes Feature Correlation')
    plt.tight_layout()
    plt.show()
```
